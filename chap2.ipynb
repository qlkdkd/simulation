# 2. 모델링과 시뮬레이션

## 2.4. 시뮬레이션 모델의 분류

### 몬테카를로 시뮬레이션

```python
import random

def pi_simulation():
    limit = 50000
    cnt = 0
    
    for i in range(0, 50001):
        u_x = random.random()
        u_y = random.random()
        if u_x**2 + u_y**2 <= 1:
            cnt += 1
    
    s = cnt / limit
    s *= 4
    print('count: ', cnt)
    print('result: ', s)

pi_simulation()
```
