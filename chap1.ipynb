## 예제 1.1

```python
import numpy as np
import matplotlib.pyplot as plt

def total_saving(total_year, salary, perint, perinc):
    init_year = 0
    savings = 0
    total_saving = []
    total_salary = []

    # total_year까지 반복
    while init_year < total_year:
        # savings 공식
        savings += (perint * savings + 0.1 * salary); total_saving.append(savings)
        # salarys 공식
        salary += perinc * salary; total_salary.append(salary)
        
        init_year += 1
    return total_salary, total_saving

salarys, savings = total_saving(30, 1200, 0.05, 0.05)
for salary, saving in zip(range(len(salarys)), range(len(savings))):
    print(f'salary {salary+1:3d}: {salarys[salary]:10.3f}, saving {saving:3d}: {savings[saving]:10.3f}')
```

## 예제 1.2

```python
from math import pi, cos, sin
import matplotlib.pyplot as plt

def ball_elasticity(x, y, v_0, angle, nbounce):
    # x, y: 초기 위치, v0: 초기속도, 
    # angle: 공이 날아가는 각도
    # nbounce: 공이 튀는 횟수
    tstep = 0.1 # 시간 증가치
    a_y = -9.8 # 중력가속도

    result_x = []
    result_y = []

    # 라디안값 변화
    angle *= (pi / 180) 
    v_x = v_0 * cos(angle)
    v_y_old = v_0 * sin(angle)
    bounce = 0 # 공이 튀어오른 수
    t = 0

    # 공 튀기면
    while bounce < nbounce:
        t += tstep
        x += v_x * tstep 
        v_y_new = v_y_old + a_y * tstep
        y += (v_y_old + v_y_new) * tstep / 2
        v_y_old = v_y_new
        result_x.append(x)
        result_y.append(y)
        print(f'x{int(10*t):3d}: {x:5.5f}, y{int(10*t):3d}: {y:5.5f}')

        if y <= 0:
            v_y_old = abs(0.75 * v_y_old)
            y = 0
            bounce += 1
    return result_x, result_y

x, y = ball_elasticity(0, 0, 10, 80, 3)

plt.plot(x, y)
plt.xlabel('x_cordinate')
plt.ylabel('y_cordinate')
plt.show()
```

# 1.3. 확률적 시뮬레이션
## 난수 발생 알고리즘

```python
def random_simulation(n):
    n *= (843314861 + 453816693)
    if n < 0:
        n += (2147483647 + 1)
    U = n * 0.4656612e-9
    return U

print(random_simulation(6))
```

```python
import random

COUNT = [0] * 6

for i in range(1, 5001):
    u = random.random()
    print(f'i={i:5}, U={u:.7f}')

    r1 = 0.0
    r2 = 0.2

    for j in range(1, 6):
        if r1 <= u < r2:
            COUNT[j] = COUNT[j] + 1

        r1 += 0.2
        r2 += 0.2

# 결과 출력 
print("\n--- 최종 구간별 카운트 결과 ---") 
for j in range(1, 6): 
    print(f"구간 {j:2d} ({(j-1)*0.2:.1f} <= U < {j*0.2:.1f}): {COUNT[j]:2d}회")
```

## 저축 시뮬레이션

```python
import random

def salarys_2(salary, perint, total_year, perhig, perlow):
    # perint: 복리 이자
    # perhig: 최대 인상률
    # perlow: 최저 인상률

    savings = 0
    year = 0

    while year < total_year:
        year += 1
        savings += (perint * savings + 0.1 * salary)
        u = random.random()

        # perinc 연봉의 인상률로 0~10% 이산형 일양확률변수
        perinc = int(u * (perhig + 1 - perlow) + perlow)
        perinc *= 0.01
        salary += perinc * salary
        print(f'year: {year:2.0f}, sal: {salary:8.6f}, savings: {savings:8.6f}')

salarys_2(12000000, 0.05, 30, 10.0, 0)
```

# 1.5. 대기행렬
## 대기행렬 문제

```python
from scipy.stats import poisson 
import random

def waiting_matrix(tlimit):
    # tlimit: 총 시뮬레이션 시간
    q = 0 # 대기행렬에서 기다리고 있는 사람(차량) 수
    totq = 0 # q의 누적, 총 대기시간
    totarr = 0 # 지금까지 도착한 총 고객 수
    tpump = 0 # 고객의 남은 봉사시간
    time = 0 # 시뮬레이션의 현재 시간
    prarr = 1/4 # 고객이 1분 이내 도착할 확률
    tstep = 1 # 시뮬레이션 단위시간(분)

    total_aveque, total_totarr = [], []

    while time < tlimit:
        time += tstep
        arrive = 0
        u = random.random()

        # 고객이 도착하면 1, 아니면 0
        if u < prarr * tstep:
            arrive = 1
            q += arrive
            totarr += 1

        # 봉사시간 > 0 : 봉사가 안끝난 경우
        if tpump > 0: 
            tpump -= tstep
            if tpump < 0:
                tpump = 0

        # 고객이 봉사를 마치고 떠났으며
        # 현재 대기행렬에 기다리고 있는 고객이 있는 경우
        if tpump == 0 and q != 0:
            q -= 1
            p = poisson.rvs(4, size=1)
            tpump = p
        totq += q

        aveque = totq / (tlimit/tstep) # 평균 대기행렬
        print(f'time: {time}, arrive: {arrive}, queue: {q}, tpump: {tpump}')
        print(f'평균 대기시간: {aveque}, 총 고객 수: {totarr},')
        print('-----------------------------------------')
        total_aveque.append(aveque)
        total_totarr.append(totarr)
    return total_aveque, total_totarr

x, y = waiting_matrix(100)
plt.plot(x, y)
plt.show()
```
