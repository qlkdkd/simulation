# 3. 대기행렬 문제
## 3.2. 복수창구의 대기행렬 시스템 문제
### 봉사창구가 2개인 경우

```python
import random
from scipy.stats import poisson

def waiting_maxtix():
    totque = 0 # 대기행렬 queue 전체 합
    aveque = 0 # 대기행렬 평균 길이
    tpump = [0] * 2 # 봉사 시간, 창구가 2개이므로 [0, 0]

    queue = 0 # 현재 대기행렬에서 기다리는 사람
    totarr = 0 # 지금까지 도착한 총 고객 수
    time = 0 # 시뮬레이션의 현재 시간
    tstep = 1 # 단위시간
    prarr = 1/3 # 고객이 1분 안에 도착할 확률
    mean = 4 # 평균이 4분인 확률변수
    tlimit = 100 # 총 시뮬레이션 수행시간

    while time < tlimit:
        time += tstep
        arrive = 0
        u = random.random()
        
        # 고객 도착여부 확인
        if u < prarr * tstep:
            arrive = 1
            queue += arrive
            totarr += 1

        # 봉사창구 개수만큼 while문 반복
        for i in range(2):
            if tpump[i] > 0:
                tpump[i] -= tstep
                if tpump[i] < 0:
                    tpump[i] = 0

            # 고객이 봉사를 마치고 떠났으며
            # 현재 대기행렬에는 기다리고 있는 고객이 있는 경우
            if tpump[i] == 0 and queue != 0:
                queue -= 1
                p = poisson.rvs(mean, 1)
                tpump[i] = p
        # end of inner for

        totque += queue
        print(f'time: {time}, arrive: {arrive}, queue: {queue}')
        print(f'tpump1: {tpump[0]}, tpump2: {tpump[1]}')
        print('---------------------------')
    # end of outer while
    aveque = totque / (tlimit/tstep)
    print(f'평균 길이: {aveque}, 총 고객 수: {totarr}')

waiting_maxtix()
```

## 3.3. 봉사시간의 변화
### 기본 문제에서 차량들이 받는 봉사시간 고려

```python
import random
import numpy as np

# 1. 초기 시뮬레이션 환경 변수 설정
prarr = 1/4       # 고객이 1분 안에 도착할 확률
mean = 4           # 평균 서비스 시간 (포아송 분포)
nmean = 20 # 정규분포 평균
stdev = 3 # 정규분포 표준편차
tlimit = 100       # 총 시뮬레이션 수행시간

# 상태 저장용 글로벌 변수
state = {
    'time': 0, # 현재 시간
    'queue': 0, # 현재 대기행렬 사람 수
    'totarr': 0, # 총 고객 수
    'tpump': 0, # 고객의 남은 봉사 시간
    'arrive': 0, # 
    'history_time': [],
    'history_queue': []
}

def waiting_matrix3():
    totque = 0 # 총 고객 수
    tstep = 1 # 시뮬레이션 단위시간
    
    # 시뮬레이션 시행
    while(state['time'] < tlimit):
        state['time'] += tstep
        state['arrive'] = 0
        u = random.random()
        if (u < (prarr * tstep)):
            state['queue'] += state['arrive']
            state['totarr'] += 1

        if (state['tpump'] > 0):
            state['tpump'] -= tstep
            if (state['tpump'] < 0):
                state['tpump'] = 0

        # 고객 봉사 완료, 현재 대기행렬 내 대기 고객 있는 경우
        if (state['tpump']==0) and (state['queue']!=0):
            state['queue'] -= 1
            liters = np.random.normal(loc=nmean, scale=stdev, size=1)
            state['tpump'] = 1 + 1 + 8*liters
        
        totque += state['queue']
        
        print(f'time: {state['time']}, arrive: {state['arrive']}, queue: {state['queue']}')
        print(f'tpump: {state['tpump']}')
        print('----------------------------------------')

    aveque = totque / (tlimit/tstep)
    print('aveque:', aveque)
    print('totarr:', state['totarr'])

waiting_matrix3()
```
